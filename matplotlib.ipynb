```python
import matplotlib.pyplot as plt
import numpy as np
from datetime import datetime, timedelta
```

```python
year = [2009, 2010, 2011, 2012, 2013]
sales = [100, 150, 200, 50, 75]
```

```python
plt.title('Analytics')
plt.xlabel('years')
plt.ylabel('sales')
plt.grid()
plt.plot(year, sales)
plt.show()
```

```python
plt.title('Analytics')
plt.xlabel('years')
plt.ylabel('sales')
plt.bar(year, sales, color='red')
plt.show()
```

```python
st_date = datetime(2025,1,1)
end_date = datetime(2026,1,1)
num_days = (end_date - st_date).days
print(st_date, num_days)
```

```python
date_list = []
for i in range(num_days):
    date_list.append(st_date.date())
    st_date += timedelta(days = 1)
```

```python
date_list
```

```python
stock_price_list = (np.random.randn(num_days)).tolist()
```

```python
stock_price_list
```

```python
np.random.seed(0)
stock_price_list = (np.random.randn(num_days).cumsum()*50).tolist()
```

```python
stock_price_list
```

```python
num = [2,5,2,5]
np.cumsum(num)
```

```python
plt.Figure(figsize=(15,6))
plt.plot(date_list,stock_price_list,color='black',label='price')
plt.title('stock price of 1 year')
plt.xlabel('date')
plt.ylabel('stock price')
# label in plot wont work without legend
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.Figure(figsize=(15,6))
plt.plot(date_list, stock_price_list, color = 'blue', marker='s', label='price')
plt.title('stock price of 1 year')
plt.xlabel('date')
plt.ylabel('stock price')
# label in plot wont work without legend
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
